# Notebook A — Baseline scenario

**Pipeline:** A (control)
**Pain Point mapped:** none (baseline / sanity check)

Reads the public NYC Yellow Taxi parquet from Azure Open Datasets, counts rows, and logs success. If this fails, the storage / network / Spark pool itself has a problem — investigate before judging the other scenarios.

Expected SRE Agent observation: pipeline succeeds, ~2M rows read, no anomalies.

```python
from pyspark.sql import SparkSession
import time

spark = SparkSession.builder.appName('notebook-A-baseline').getOrCreate()

url = 'wasbs://nyctlc@azureopendatastorage.blob.core.windows.net/yellow/puYear=2019/puMonth=1/'
print(f'[notebook-A-baseline] Reading: {url}')
start = time.time()

df = spark.read.parquet(url)
row_count = df.count()
col_count = len(df.columns)
elapsed = round(time.time() - start, 1)

print(f'[notebook-A-baseline] rowsRead={row_count} colsRead={col_count} durationSec={elapsed}')
print(f'[notebook-A-baseline] columns={df.columns}')
print(f'[notebook-A-baseline] Status: SUCCEEDED')
```
